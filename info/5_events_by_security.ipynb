```python
url = "https://iss.moex.com/iss/securities/SBER/events.json"
# "https://iss.moex.com/iss/securities/{SECID}/events.json"
# Корпоративные действия и дивиденды - Фундаментальные данные
# Полную историю корпоративных событий по инструменту - дивиденды, сплиты, собрания акционеров и другие фундаментальные данные.

response = requests.get(url)
data = response.json()
data.keys()
```

```python
df = pd.DataFrame(data["boards"]["data"], columns=data["boards"]["columns"])
```
