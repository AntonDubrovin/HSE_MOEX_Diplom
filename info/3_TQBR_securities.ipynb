```python
import requests
import pandas as pd
import json
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
url = "https://iss.moex.com/iss/engines/stock/markets/shares/boards/TQBR/securities.json"
# Текущие рыночные данные - Реал-тайм мониторинг
# Актуальные данные по ВСЕМ инструментам основного режима торгов в реальном времени.

response = requests.get(url)
data = response.json()
data.keys()
```

```python
df = pd.DataFrame(data["securities"]["data"], columns=data["securities"]["columns"])
```

Детальные данные (marketdata):

Столбец	|Тип|	Описание
| --- | --- | --- |
SECID	|String	|Код инструмента. Идентификатор
BOARDID	|String	|Режим торгов. Идентификация режима
SHORTNAME	|String	|Краткое название
PREVPRICE	|Decimal|	Цена предыдущего закрытия. Базис для изменений
LOTSIZE	|Integer	|Размер лота. Расчет стоимости лота
FACEVALUE|	Decimal	|Номинальная стоимость. Фундаментальный показатель
STATUS	|String|	Статус инструмента. Доступность торгов
DECIMALS|	Integer	|Количество знаков после запятой. Точность цены
SECNAME	|String	|Название инструмента. Полное наименование
MINSTEP	|Decimal|	Минимальный шаг цены. Для точных расчетов
PREVWAPRICE|	Decimal|	Предыдущая WAP цена. Для анализа
FACEUNIT|	String	|Валюта номинала. Валюта инструмента
PREVDATE|	Date	|Дата предыдущего закрытия. Временная метка
ISSUESIZE|	Integer	|Объем выпуска. Анализ капитализации
ISIN	|String	|ISIN код. Международный идентификатор
LISTLEVEL	|Integer|	Уровень листинга. Качество бумаги
CURRENCYID	|String	|Валюта расчета. Валюта торгов
