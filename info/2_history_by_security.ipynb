```python
import requests
import pandas as pd
import json
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
url = "https://iss.moex.com/iss/history/engines/stock/markets/shares/securities/SBER.json"
# "https://iss.moex.com/iss/history/engines/stock/markets/shares/securities/{SECID}.json"
# Исторические данные по инструменту - Основные временные ряды
# Полные исторические данные по конкретному инструменту - основа для любого финансового анализа.

response = requests.get(url)
data = response.json()
data.keys()
```

```python
df = pd.DataFrame(data["history"]["data"], columns=data["history"]["columns"])
```

Детальные данные:
Столбец	|Тип	|Описание
| --- | --- | --- |
TRADEDATE	|Date	|Дата торгов. Для временных рядов и анализа
SECID	|String |Код инструмента. Идентификатор
OPEN	|Decimal	|Цена открытия. Анализ гэпов, волатильности открытия
CLOSE	|Decimal	|Цена закрытия. Главный показатель для индикаторов
HIGH	|Decimal	|Максимальная цена дня. Уровни сопротивления, волатильность
LOW	|Decimal	|Минимальная цена дня. Уровни поддержки
VOLUME	|Integer	|Объем в штуках. Подтверждение тренда, ликвидность
VALUE	|Decimal	|Оборот в рублях. Денежный объем, интерес крупных игроков
NUMTRADES	|Integer	|Количество сделок. Активность трейдеров
WAPRICE	|Decimal	|Средневзвешенная цена. Более точная цена для анализа
LEGALCLOSEPRICE	|Decimal	|Цена закрытия для юр.лиц
CURRENCYID	|String	|Валюта расчета. Валютные риски
BOARDID	|String	|Режим торгов. Идентификация площадки
TRENDCLSPR	|Decimal	|Изменение цены закрытия. Моментный тренд
TRADE_SESSION_DATE	|Date	|Дата торговой сессии. Техническая информация
