```python
import requests
import pandas as pd
import json
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
url = "https://iss.moex.com/iss/securities.json"
# Полный каталог ВСЕХ финансовых инструментов Московской биржи.

response = requests.get(url)
data = response.json()
data.keys()
```

```python
df = pd.DataFrame(data["securities"]["data"], columns=data["securities"]["columns"])
```

Детальные данные:

Столбец	|Тип	| Описание
| --- | --- | --- |
secid	|String	| Уникальный код инструмента. Основной идентификатор для всех запросов
shortname	|String	| Краткое название. Для отображения и фильтрации
name	|String	|Полное наименование. Инфо
isin	|String	|Международный код ISIN. Для международной идентификации
is_traded	|Boolean	|Флаг торгуемости. Фильтр активных инструментов
emitent_id | Integer | ID эмитента. связь с эмитентом
emitent_title	|String	|Наименование эмитента. Анализ по компаниям
emitent_inn	|String	|ИНН эмитента. Юридическая идентификация
emitent_okpo | String | Доп код
type	|String	|Тип инструмента. Классификация (акции облигации итд)
group	|String	|Группа инструментов. Отраслевая классификация
primary_boardid	|String	|Основной режим торгов. Последующие запросы данных
marketprice_boardid	|String	|Режим для расчета цен. Инфо
