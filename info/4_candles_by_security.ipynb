```python
import requests
import pandas as pd
import json
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
url = "https://iss.moex.com/iss/engines/stock/markets/shares/securities/SBER/candles.json"
# "https://iss.moex.com/iss/engines/stock/markets/shares/securities/{SECID}/candles.json"
# Свечные данные - Агрегированные временные ряды
# Агрегированные данные за указанный временной интервал - идеально для технического анализа.

response = requests.get(url)
data = response.json()
data.keys()
```

```python
df = pd.DataFrame(data["candles"]["data"], columns=data["candles"]["columns"])
```

Детальные данные:
Столбец	    |Тип	|Описание
| --- | --- | --- |
open	    |Decimal	|Цена открытия
close	    |Decimal	|Цена закрытия
high	    |Decimal	|Максимальная цена
low	        |Decimal	|Минимальная цена
value	    |Decimal	|Оборот в рублях
volume	    |Integer	|Объем в штуках
begin	    |DateTime	|Начало свечи
end	        |DateTime	|Конец свечи
